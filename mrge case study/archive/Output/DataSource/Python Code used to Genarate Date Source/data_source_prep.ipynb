## Joins - identify appropriate joins and connect the tables

### A Star schema

```python
# fact table has ids as keys for all the dimension tables
# join the tables to create a final sql query - that can be used to create a view/table to be fed to Tableau datasource
# also get the results of final resulting table in an excel that can be connected to Tableau datasource

conn = sqlite3.connect("retail_data.db")
```

```python
conn.execute("""
DROP VIEW IF EXISTS view_retail_joined_dataset;
""")
conn.commit()
```

```python
# order is a reserved statement in sql - can't use it as alias
# names of columns are already formatted properly - so no need to rename as they all look good for datasource consumption
# from my experience working with datasets - IDs are relevant in exposing datasets as they can be used for counts in Tableau datasource
# create counts in Tableau data source using ids - and then hide ids? - Specifically product id and customer id can be hiden in Tableau data source
# zip - as per my experience good for exposure because of very granular map visuals in Tableau. Hence, keeping it

# profit / revenue: derived metric - better in Tableau as if done here - taleau will confuse with reaggregations

# using CTE to only demonstrate that my personal preference as matter of best practice is to use them!

conn.execute("""

CREATE VIEW view_retail_joined_dataset AS 

WITH base AS (
    SELECT
        /* ---------- Product fields ---------- */
        product."Product ID",
        product."Product Type",
        product."Sub-Category",
        product."Product Name",

        /* ---------- Customer fields ---------- */
        customer."Customer ID",
        customer."Customer Name",
        customer."Segment",
        customer."Country/Region",
        customer."City",
        customer."State/Province",
        customer."Postal Code",
        customer."Region",

        /* ---------- Order fields ---------- */
        orders."Order ID",
        orders."Order Date",
        orders."Ship Date",
        orders."Ship Mode",
        CASE 
            WHEN orders."Is Returned" = 0 THEN 'Not Returned'
            WHEN orders."Is Returned" = 1 THEN 'Returned'
        ELSE 'NA' 
        END AS "Is Returned",

        /* ---------- Sales metrics ---------- */
        sales."Price Per Unit",
        sales.Quantity,
        sales.Discount,
        sales.Profit,

        /* ---------- Derived metrics ---------- */
        sales."Price Per Unit"
            * sales.Quantity
            * (1 - sales.Discount) AS Revenue,

        CAST(
            julianday(orders."Ship Date")
            - julianday(orders."Order Date")
            AS INTEGER
        ) AS "Shipping Days"

    FROM fact_sales AS sales

    -- Join product dimension
    LEFT JOIN dim_product AS product
        ON sales."Dim Product ID" = product."Dim Product ID"

    -- Join customer dimension
    LEFT JOIN dim_customer AS customer
        ON sales."Dim Customer ID" = customer."Dim Customer ID"

    -- Join order dimension
    LEFT JOIN dim_order AS orders
        ON sales."Dim Order ID" = orders."Dim Order ID"
)

SELECT
    *
FROM base
""")
```

```python
conn.commit()
```
